```python
#setup game
import random
#generate a number between 1 and 10
secret_num=random.randint(1,10)
```

```python
##### Prompt the User, Set a variable attempts ,maximum 3 using while loop,exit if right or continue with conditions


attempts=0
max_attempts=3

#prompt user to input value ,while maximum attempts are 3
while attempts < 3:
    num=int(input("guess the number between(1 and 10):"))
    
#using continue ,to skip this attempt from maximum count, if value is out of range    
    if num < 1 or num >10:
        print("guess is out of range.try number between 1 and 10")
        continue
        
    attempts += 1
    
#conditions and feedback,break,when guess is right    
    if num ==secret_num:
       print("congratulations!! you won the game")
       break
    
    elif num < secret_num:
        print("too low,try again")
    else:
        print("too high,try again")
    
else:
    print("your attempts reached maximum for today,better luck next time!")
```

                      BMI Calculator

```python
def calculate_bmi(weight,height):
    bmi = weight / (height**2)
    return bmi

weight=float(input("enter your weight in kg:"))
height=float(input("enter your height in metre:"))
bmi=calculate_bmi(weight,height)
print("Your BMI is:",bmi)
```
